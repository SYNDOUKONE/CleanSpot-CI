# 04 — Segmentation des amas (YOLO-Seg / Mask R-CNN / U-Net)

Poids : `models/segmentation/best-seg.pt` (YOLO-Seg).

```python
from ultralytics import YOLO

seg = YOLO("yolov8n-seg.pt")
seg.train(
    data="../datasets/taco/yolo_cleanspot/dataset.yaml",
    epochs=80,
    imgsz=640,
    task="segment",
    name="cleanspot-seg",
)
```

```python
import numpy as np
res = seg.predict("../datasets/cleanspot_ci/images/exemple.jpg")[0]
masks = res.masks.data.cpu().numpy() if res.masks is not None else []
soiled = float(sum(m.sum() for m in masks)) / masks[0].size if masks else 0.0
print(f"surface souillée : {soiled:.2%}")
```
