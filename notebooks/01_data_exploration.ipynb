```python
import json
from collections import Counter
from pathlib import Path

TACO = Path("..").resolve() / "datasets" / "taco"
data = json.loads((TACO / "annotations.json").read_text())
print(f"{len(data['images'])} images, {len(data['annotations'])} annotations")
```

## Répartition par catégorie (TACO)

```python
categories = {c["id"]: c for c in data["categories"]}
sc = Counter(categories[a["category_id"]]["name"] for a in data["annotations"])
sc.most_common(15)
```

```python
import matplotlib.pyplot as plt
labels, values = zip(*sc.most_common(15))
plt.figure(figsize=(10, 5))
plt.barh(labels[::-1], values[::-1])
plt.title("Top 15 catégories TACO")
plt.tight_layout()
```
