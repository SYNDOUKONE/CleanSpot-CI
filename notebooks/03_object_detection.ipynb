# 03 — Détection d'objets (YOLOv8n / YOLO11n)

Entraînement sur TACO + pLitterStreet + CleanSpot CI. Poids de sortie : `models/detection/best.pt` (format YOLO).

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")  # ou yolo11n.pt
results = model.train(
    data="../datasets/taco/yolo_cleanspot/dataset.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    name="cleanspot-detection",
)
```

```python
metrics = model.val()
print(metrics.box.map50, metrics.box.map)
```
